```python
import torch
import torch.nn as nn

vocab_size = 8
embed_dim = 4
embedding = nn.Embedding(vocab_size, embed_dim)

custom_weights = torch.tensor( [[-0.1882,  0.5530,  1.6267,  0.7013],
                                [ 1.7840, -0.8278, -0.2701,  1.3586],
                                [ 1.0281, -1.9094,  0.3182,  0.4211],
                                [-1.3083, -0.0987,  0.7647, -0.3680],
                                [ 0.2293,  1.3255,  0.1318,  2.0501],
                                [ 0.4058, -0.6624, -0.8745,  0.7203],      
                                [ 0.5582,  0.0786, -0.6817,  0.6902],
                                [ 0.4309, -1.3067, -0.8823,  1.5977]]).float()
embedding.weight = nn.Parameter(custom_weights)
print(embedding.weight)
```

```python
data = torch.tensor([0, 4, 7, 2, 1], dtype = torch.long)
data_embedding = embedding(data)
print(f'Data Embedding: {data_embedding}')
```

```python
embed_dim = 4
hidden_dim = 3
rnn = nn.RNN(
    embed_dim, hidden_dim, batch_first = True
)
print(rnn)
total_params = 0
for name, param in rnn.named_parameters():
    print(name, param.shape, param.numel())
    total_params += param.numel()
print('Total params: ', total_params)

output, hn = rnn(data_embedding)
print(f'output shape: {output.shape}')
print(f'hn shape: {hn.shape}')
```

```python
vocab_size = 8
embed_dim = 4
embedding = nn.Embedding(vocab_size, embed_dim)

init_weights = torch.tensor( [[-0.1882,  0.5530,  1.6267,  0.7013],
                                [ 1.7840, -0.8278, -0.2701,  1.3586],
                                [ 1.0281, -1.9094,  0.3182,  0.4211],
                                [-1.3083, -0.0987,  0.7647, -0.3680],
                                [ 0.2293,  1.3255,  0.1318,  2.0501],
                                [ 0.4058, -0.6624, -0.8745,  0.7203],
                                [ 0.5582,  0.0786, -0.6817,  0.6902],
                                [ 0.4309, -1.3067, -0.8823,  1.5977]]).float()
embedding.weight = nn.Parameter(init_weights)
data = torch.tensor([0, 4, 7, 2, 1])
data_embedding = embedding(data)
print(data_embedding)
```

```python
embed_dim = 4
hidden_dim = 3
rnn = nn.RNN(
    embed_dim,
    hidden_dim,
    num_layers = 1,
    batch_first = True
)
rnn.bias_hh_l0 = nn.Parameter(torch.tensor([0.4481,  0.5537, -0.5006], dtype = torch.float))
rnn.bias_ih_l0 = nn.Parameter(torch.tensor([0.0135, -0.2209,  0.1330], dtype = torch.float32))

rnn.weight_ih_l0 = nn.Parameter(torch.tensor([
    [-0.4174,  0.1953, -0.0365, -0.4025],
    [ 0.4722, -0.4085, -0.0236,  0.3763],
    [ 0.0550, -0.4921, -0.4307,  0.0855]
], dtype = torch.float32))

rnn.weight_hh_l0 = nn.Parameter(torch.tensor([
    [-0.5511,  0.1260,  0.0463],
    [-0.2291, -0.2084, -0.2352],
    [-0.4341, -0.2682,  0.0612]], 
    dtype = torch.float32))
```

```python
test_data = data_embedding.reshape(1, 5, 4)
print(test_data.shape)
```

```python
output, hn = rnn(test_data)

# (N, L, H_out)
print(output.shape)
print(output)
print(output[:, -1, :])
```

```python
# (Num_layers, N, H_out)
print(hn.shape)
print(hn)
print(hn[-1, :, :])
```
